```python
Nama  = "Budi"
Umur  = 18
Kelas = "9F"

def Data_Siswa(Nama,Umur,Kelas):
    print(f"Nama : {Nama}\nUmur : {Umur}\nKelas: {Kelas}")

Data_Siswa(Nama,Umur,Kelas)
```

### • Classes

Kelas merupakan sebuah template atau format yang mendefinisikan metode dan parameter apa saja yang terikat ke dalam objek tertentu, misalnya:

```python
class Data_Siswa:
    def __init__(self,Nama,Umur,Kelas):
        self.Nama = Nama
        self.Umur = Umur
        self.Kelas= Kelas

    def tampilkan_data(self):
        print(f"Nama : {self.Nama}\nUmur : {self.Umur}\nKelas: {self.Kelas}")

Data_Siswa
```

dari kode tersebut terlihat bahwa suatu kelas dengan nama 'Data_siswa' telah berhasil dibuat yang dibuktikan dengan hasil output dari script baris terakhir yang menjelaskan bahwa 'Data_Siswa' telah tersimpan di modul `__main__`, selanjutnya untuk menampilkan keterangan dari sebuah class dapat digunakan fungsi help( ) seperti berikut:

```python
help(Data_Siswa)
```

### • Methods

Metode merupakan suatu fungsi yang didefinisikan dalam sebuah kelas dan berfungsi untuk memanipulasi suatu objek dengan tugas tertentu, misalnya:

```python
class Data_Siswa:
    def __init__(self,Nama,Umur,Kelas):
        self.Nama = Nama
        self.Umur = Umur
        self.Kelas= Kelas

    def tampilkan_data(self):
        print(f"Nama : {self.Nama}\nUmur : {self.Umur}\nKelas: {self.Kelas}")

Siswa_1 = Data_Siswa("Budi",18,"9F")

Siswa_1.tampilkan_data()
```

pada kode di atas kelas 'Data_Siswa' memiliki 2 metode, yaitu:
1. metode `__init__` dengan 4 parameter (self,Nama,Umur,Kelas) di mana 3 nilai dari parameter tersebut disimpan sebagai atribut milik parameter self,
2. Metode tampilkan_data memiliki 1 parameter (self) yang digunakan untuk mengakses dan mencetak ketiga atribut objek tersebut ke layar.

### • Attributes

Atribut merupakan variabel dari beberapa kumpulan data dengan nilai tertentu yang melekat pada suatu parameter dalam sebuah kelas, misalnya:

```python
class Data_Siswa:
    def __init__(self,Nama,Umur,Kelas):
        self.Nama = Nama
        self.Umur = Umur
        self.Kelas= Kelas

    def tampilkan_data(self):
        print(f"Nama : {self.Nama}\nUmur : {self.Umur}\nKelas: {self.Kelas}")
```

pada pembahasan sebelumnya diketahui bahwa nilai dari parameter Nama, Umur, dan Kelas akan disimpan sebagai atribut dari parameter self sehingga self memiliki 3 variabel yang melekat kepadanya yaitu:

```python
Siswa_1 = Data_Siswa("Budi",18,"9F")
```

```python
Siswa_1.Nama
```

```python
Siswa_1.Umur
```

```python
Siswa_1.Kelas
```

#### - Magic Methods (Dunder Methods)

**Dunder Methods** atau singkatan dari *"double underscore methods"* merupakan adalah sekumpulan fungsi rahasia yang berjalan secara otomatis ketika sebuah objek berinteraksi dengan sintaks dasar Python. *Dunder* disebut juga sebagai *magic method* karena kita tidak perlu memanggil namanya secara langsung agar fungsinya bekerja; mereka bereaksi secara gaib terhadap simbol atau kondisi tertentu. Misalnya:

metode __add__ dapat dituliskan sebagai

```python
str.__add__('halo',' semua')
```

atau dapat juga ditulis sebagai

```python
'halo'.__add__(' semua')
```

beberapa *Dunder Methods* seperti `__add__` dapat digantikan dengan `operator`, misalnya operator `+`

```python
'halo' + ' semua'
```

beberapa `*Dunder Methods*` lainnya juga ada yang terdapat dalam sebuah fungsi, seperti `len()`

```python
text = 'Halo Semua!'
```

```python
text.__len__()
```

dapat digantikan dengan

```python
len(text)
```
